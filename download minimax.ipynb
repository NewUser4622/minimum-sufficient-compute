```python
import os
import sys
import time
from huggingface_hub import hf_hub_download, login

# Enable Rust multi-threaded fast transfer engine
os.environ["HF_HUB_ENABLE_HF_TRANSFER"] = "1"

# ==============================================================================
# CONFIGURATION
# ==============================================================================
HF_TOKEN = "token here"
SAVE_DIR = r"C:\MiniMax_H3_Weights"

if HF_TOKEN and HF_TOKEN.startswith("hf_"):
    login(token=HF_TOKEN)

REPO_ID = "Comfy-Org/MiniMax-H3"
FILES_TO_DOWNLOAD = [
    "diffusion_models/minimax_h3_fl2va_pruned_int8_convrot.safetensors",
    "text_encoders/qwen3vl_32b_minimax_h3_nvfp4_awq.safetensors",
    "vae/minimax_h3_video_vae_fp16.safetensors",
    "vae/minimax_h3_audio_vae_fp32.safetensors",
]

os.makedirs(SAVE_DIR, exist_ok=True)

print("================ RUST-ACCELERATED DOWNLOAD MANAGER ================")
print(f"Target Directory: {SAVE_DIR}\n")

for repo_file in FILES_TO_DOWNLOAD:
    target_path = os.path.join(SAVE_DIR, repo_file.replace("/", os.sep))
    
    if os.path.exists(target_path):
        size_gb = os.path.getsize(target_path) / (1024**3)
        print(f"[EXISTS] {repo_file} ({size_gb:.2f} GB)")
        continue
        
    success = False
    max_retries = 5
    attempt = 0
    
    while not success and attempt < max_retries:
        attempt += 1
        try:
            print(f"\n[DOWNLOADING] {repo_file} (Attempt {attempt}/{max_retries})")
            sys.stdout.flush()  # Force instant UI refresh
            
            hf_hub_download(
                repo_id=REPO_ID,
                filename=repo_file,
                local_dir=SAVE_DIR,
                token=HF_TOKEN if HF_TOKEN.startswith("hf_") else None,
                local_dir_use_symlinks=False
            )
            print(f"[COMPLETED] {repo_file}\n")
            success = True
        except Exception as e:
            print(f"[ERROR] Connection/DNS interruption: {e}")
            if attempt < max_retries:
                print("Retrying in 5 seconds...")
                time.sleep(5)
            else:
                print(f"[FAILED] Could not download {repo_file} after {max_retries} attempts.")
                sys.exit(1)

print("\n================ All Downloads Complete! ================")
```
